```python
import numpy as np
```

```python
# columns = [age, heart_rate, cholesterol, blood_pressure]
patients = np.array([
    [45, 80, 200, 120],
    [60, 90, 240, 140],
    [30, 70, 180, 110],
    [50, 85, 220, 130],
    [65, 95, 260, 150]
])
```

```python
np.mean(patients[:, 0])   #average age
```

```python
np.max(patients[:, 1])   #maximum heart rate
```

```python
np.min(patients[:, 2])    #minimum cholesterol level
```

```python
patients[patients[:, 3] > 130]  #blood pressure > 130
```

```python
patients[patients[:, 2] >= 240]   # cholesterol ≥ 240
```

```python
patients[patients[:, 0] > 50]   #older than 50
```

```python
patients[(patients[:, 3] > 140) | (patients[:, 2] > 240)]  #high-risk patients
```

```python
under50_avg = np.mean(patients[patients[:, 0] < 50][:, 1])
over50_avg = np.mean(patients[patients[:, 0] >= 50][:, 1])

print(under50_avg)
print(over50_avg)
```

```python
patients[patients[:, 3] == max(patients[:, 3])]  #Highest blood pressure
```

```python
patients[patients[:, 1] == min(patients[:, 1])]    #Lowest heart rate
```

```python
print(patients[:, [0, 3]])   # age vs blood pressure

print("Older patients tend to have higher blood pressure (positive relationship in this dataset)")
```
