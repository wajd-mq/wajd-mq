```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
import pandas as pd
import time

# open browser
driver = webdriver.Chrome()
driver.get("https://timesofoman.com/")
time.sleep(5)  # wait for page to load

# lists
titles = []
dates = []
images = []

# find article cards (IMPORTANT: may need adjustment)
articles = driver.find_elements(By.XPATH, "//article")[:30]

for article in articles:
    try:
        # TITLE
        title = article.find_element(By.XPATH, ".//h2 | .//h3").text
    except:
        title = None

    try:
        # DATE
        date = article.find_element(By.XPATH, ".//time").text
    except:
        date = None

    try:
        # IMAGE URL
        img = article.find_element(By.XPATH, ".//img").get_attribute("src")
    except:
        img = None

    titles.append(title)
    dates.append(date)
    images.append(img)

driver.quit()

# create table
df = pd.DataFrame({
    "Title": titles,
    "Date": dates,
    "Image": images
})

print(df)

# save to CSV
df.to_csv("times_of_oman.csv", index=False)
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time

driver = webdriver.Chrome()
driver.get("https://timesofoman.com/")
time.sleep(6)

titles = []
dates = []
images = []

# 1. Collect article links first
links = driver.find_elements(By.XPATH, "//a[contains(@href, '/article') or contains(@href, '/news')]")

urls = []

for link in links:
    url = link.get_attribute("href")
    if url and url not in urls:
        urls.append(url)

# limit to avoid overload
urls = urls[:30]

# 2. Visit each article page
for url in urls:
    try:
        driver.get(url)
        time.sleep(3)

        # TITLE
        try:
            title = driver.find_element(By.XPATH, "//h1").text
        except:
            title = None

        # DATE
        try:
            date = driver.find_element(By.XPATH, "//time").text
        except:
            date = "N/A"

        # IMAGE
        try:
            image = driver.find_element(By.XPATH, "//img").get_attribute("src")
        except:
            image = "N/A"

        titles.append(title)
        dates.append(date)
        images.append(image)

    except:
        continue

driver.quit()

# 3. Create DataFrame
df = pd.DataFrame({
    "Title": titles,
    "Date": dates,
    "Image": images
})

print(df)
df.to_csv("times_of_oman_articles.csv", index=False)
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time

driver = webdriver.Chrome()
driver.get("https://timesofoman.com/")
time.sleep(6)

titles = []
dates = []
images = []
urls = set()

# 1. collect ONLY meaningful links (must have text)
all_links = driver.find_elements(By.XPATH, "//a")

for a in all_links:
    try:
        url = a.get_attribute("href")
        text = a.text.strip()

        # filter real news links
        if url and text and len(text) > 40:
            urls.add(url)

    except:
        continue

urls = list(urls)[:30]

# 2. visit each article
for url in urls:
    try:
        driver.get(url)
        time.sleep(3)

        # TITLE (strong selector)
        try:
            title = driver.find_element(By.XPATH, "//h1").text
        except:
            continue  # skip if no title → not article

        # DATE (meta fallback - IMPORTANT)
        try:
            date = driver.find_element(By.XPATH, "//time").text
            if not date:
                date = "N/A"
        except:
            date = "N/A"

        # IMAGE (BEST METHOD)
        try:
            image = driver.find_element(By.XPATH, "//meta[@property='og:image']").get_attribute("content")
        except:
            image = "N/A"

        titles.append(title)
        dates.append(date)
        images.append(image)

    except:
        continue

driver.quit()

df = pd.DataFrame({
    "Title": titles,
    "Date": dates,
    "Image": images
})

print(df)
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time

driver = webdriver.Chrome()

driver.get("https://timesofoman.com/")
time.sleep(5)

titles = []
dates = []
images = []
article_urls = set()

# جمع روابط الأخبار
links = driver.find_elements(By.TAG_NAME, "a")

for link in links:
    try:
        url = link.get_attribute("href")
        text = link.text.strip()

        if url and text and len(text) > 40:
            article_urls.add(url)

    except:
        pass

# أول 30 مقال
article_urls = list(article_urls)[:30]

for url in article_urls:
    try:
        driver.get(url)
        time.sleep(2)

        # العنوان
        try:
            title = driver.find_element(By.TAG_NAME, "h1").text
        except:
            continue

        # التاريخ
        try:
            date = driver.find_element(By.TAG_NAME, "time").text
        except:
            date = "N/A"

        # الصورة
        try:
            image = driver.execute_script("""
                return document.querySelector(
                    'meta[property="og:image"]'
                )?.content;
            """)
        except:
            image = "N/A"

        titles.append(title)
        dates.append(date)
        images.append(image)

        print("Saved:", title)

    except Exception as e:
        print("Error:", e)

driver.quit()

df = pd.DataFrame({
    "Title": titles,
    "Date": dates,
    "Image": images
})

df.to_csv("times_of_oman_news.csv", index=False)

print(df.head())
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time

# فتح المتصفح
driver = webdriver.Chrome()

# فتح الموقع
driver.get("https://timesofoman.com/")
time.sleep(5)

titles = []
dates = []
images = []

# جمع الروابط
urls = set()

links = driver.find_elements(By.TAG_NAME, "a")

for link in links:
    try:
        url = link.get_attribute("href")
        text = link.text.strip()

        if url and text and len(text) > 40:
            urls.add(url)

    except:
        continue

# أول 30 رابط
urls = list(urls)[:30]

print("Number of links found:", len(urls))

# زيارة كل رابط
for url in urls:

    try:
        driver.get(url)
        time.sleep(2)

        # العنوان
        try:
            title = driver.find_element(By.TAG_NAME, "h1").text
        except:
            continue

        # التاريخ
        try:
            date = driver.execute_script("""
            return document.querySelector(
            'meta[property="article:published_time"]'
            )?.content;
            """)
        except:
            date = "N/A"

        if not date:
            date = "N/A"

        # الصورة
        try:
            image = driver.execute_script("""
            return document.querySelector(
            'meta[property="og:image"]'
            )?.content;
            """)
        except:
            image = "N/A"

        if not image:
            image = "N/A"

        titles.append(title)
        dates.append(date)
        images.append(image)

        print("Collected:", title)

    except:
        continue

# إغلاق المتصفح
driver.quit()

# إنشاء DataFrame
df = pd.DataFrame({
    "Title": titles,
    "Date": dates,
    "Image": images
})

# طباعة النتائج
print("\nFinal Data:")
print(df.to_string(index=False))
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time

# فتح المتصفح
driver = webdriver.Chrome()

# فتح الموقع
driver.get("https://timesofoman.com/")
time.sleep(5)

titles = []
dates = []
images = []

# جمع روابط المقالات
urls = set()

links = driver.find_elements(By.TAG_NAME, "a")

for link in links:
    try:
        url = link.get_attribute("href")
        text = link.text.strip()

        if url and text and len(text) > 40:
            urls.add(url)

    except:
        continue

urls = list(urls)[:30]

print("Found", len(urls), "articles")

# زيارة كل مقال
for url in urls:

    try:
        driver.get(url)
        time.sleep(3)

        # العنوان
        try:
            title = driver.find_element(By.TAG_NAME, "h1").text
        except:
            continue

        # التاريخ
        date = "N/A"
        try:
            elements = driver.find_elements(By.XPATH, "//*")

            for e in elements:
                text = e.text.strip()

                if ("AM" in text or "PM" in text) and len(text) < 50:
                    date = text
                    break

        except:
            pass

        # الصورة
        try:
            image = driver.execute_script("""
                return document.querySelector(
                    'meta[property="og:image"]'
                )?.content;
            """)
        except:
            image = "N/A"

        titles.append(title)
        dates.append(date)
        images.append(image)

        print("Collected:", title)

    except:
        continue

# إغلاق المتصفح
driver.quit()

# عرض النتائج
df = pd.DataFrame({
    "Title": titles,
    "Date": dates,
    "Image": images
})

print("\n================ RESULTS ================\n")
print(df.to_string(index=False))
```
