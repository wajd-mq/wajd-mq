```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree

# 1. تحميل مجموعة بيانات Iris
iris = load_iris()
X = iris.data
y = iris.target

# 2. تدريب مصنف شجرة القرار
# نستخدم max_depth=None للسماح للشجرة بالنمو الكامل (بدون تقليم مبدئي)
# random_state للتحكم في العشوائية لضمان تكرار النتائج
clf = DecisionTreeClassifier(random_state=0)
clf.fit(X, y)

# 3. رسم شجرة القرار
plt.figure(figsize=(15, 10)) # لتكبير حجم الرسمة لتكون واضحة
plot_tree(clf,
          feature_names=iris.feature_names, # أسماء الميزات (sepal length, petal width, etc.)
          class_names=iris.target_names,   # أسماء الفئات (setosa, versicolor, virginica)
          filled=True,                     # تلوين العقد بناءً على الفئة الأكثر شيوعًا
          rounded=True,                    # حواف مستديرة للعقد
          proportion=False,                # عرض عدد العينات وليس نسبة مئوية
          fontsize=10)                     # حجم الخط داخل العقد
plt.title("Decision tree trained on all the iris features", fontsize=16)
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.tree import DecisionTreeRegressor, plot_tree

# 1. تحميل مجموعة بيانات السكري (Diabetes Dataset) للانحدار
diabetes = load_diabetes()
X = diabetes.data
y = diabetes.target

# 2. تدريب مصنف شجرة القرار للانحدار (DecisionTreeRegressor)
# نستخدم max_depth للحد من تعميق الشجرة ومنع overfitting
# random_state للتحكم في العشوائية لضمان تكرار النتائج
regressor = DecisionTreeRegressor(max_depth=3, random_state=0) # تم إضافة max_depth=3 هنا
regressor.fit(X, y)

# 3. رسم شجرة القرار للانحدار
plt.figure(figsize=(20, 15)) # لتكبير حجم الرسمة لتكون واضحة
plot_tree(regressor,
          feature_names=diabetes.feature_names, # أسماء الميزات
          filled=True,                           # تلوين العقد بناءً على قيمة الانحدار
          rounded=True,                          # حواف مستديرة للعقد
          fontsize=8)                           # حجم الخط داخل العقد
plt.title("Decision Tree Regressor trained on the Diabetes dataset (max_depth=3)", fontsize=16)
plt.show()
```

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor

# ----------------------------
# Create a random dataset
# ----------------------------
rng = np.random.RandomState(1)

X = np.sort(5 * rng.rand(80, 1), axis=0)
y = np.sin(X).ravel()

# Add noise to every 5th sample
y[::5] += 3 * (0.5 - rng.rand(16))

# ----------------------------
# Train Decision Tree Regression models
# ----------------------------
regr_1 = DecisionTreeRegressor(max_depth=2, random_state=0)
regr_2 = DecisionTreeRegressor(max_depth=5, random_state=0)

regr_1.fit(X, y)
regr_2.fit(X, y)

# ----------------------------
# Prediction
# ----------------------------
X_test = np.arange(0.0, 5.0, 0.01)[:, np.newaxis]

y_1 = regr_1.predict(X_test)
y_2 = regr_2.predict(X_test)

# ----------------------------
# Plot the results
# ----------------------------
plt.figure(figsize=(9,5))

plt.scatter(
    X,
    y,
    s=20,
    edgecolor="black",
    c="darkorange",
    label="data"
)

plt.plot(
    X_test,
    y_1,
    color="cornflowerblue",
    linewidth=2,
    label="max_depth=2"
)

plt.plot(
    X_test,
    y_2,
    color="yellowgreen",
    linewidth=2,
    label="max_depth=5"
)

plt.xlabel("data")
plt.ylabel("target")
plt.title("Decision Tree Regression")
plt.legend()

plt.show()
```
