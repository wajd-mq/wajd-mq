```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Score': [45, 50, 52, 48, 51, 49, 47, 53, 46, 120]
})

mean = df['Score'].mean()
std = df['Score'].std()

df['Z_Score'] = (df['Score'] - mean) / std

print(df)
```

```python
outliers = df[abs(df['Z_Score']) > 2]

print(outliers)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Create dataset
df = pd.DataFrame({
    "Age": [22,25,28,35,45,52,23,40,30,48],
    "Salary": [25000,32000,45000,58000,85000,95000,27000,72000,50000,91000],
    "Website_Time": [5,12,20,25,40,35,7,30,18,38],
    "Distance": [1.2,3.5,5.1,8.0,15.5,20.0,2.0,10.5,6.0,18.0]
})

# Standardization
scaler = StandardScaler()

scaled_features = scaler.fit_transform(df)

scaled_df = pd.DataFrame(
    scaled_features,
    columns=df.columns
)

print(scaled_df)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Create dataset
df = pd.DataFrame({
    "Age": [22,25,28,35,45,52,23,40,30,48],
    "Salary": [25000,32000,45000,58000,85000,95000,27000,72000,50000,91000],
    "Website_Time": [5,12,20,25,40,35,7,30,18,38],
    "Distance": [1.2,3.5,5.1,8.0,15.5,20.0,2.0,10.5,6.0,18.0]
})

# Standardization
scaler = StandardScaler()

scaled_features = scaler.fit_transform(df)

scaled_df = pd.DataFrame(
    scaled_features,
    columns=df.columns
)

print(scaled_df)
```
