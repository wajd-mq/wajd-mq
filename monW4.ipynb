```python
pationt_ID = np.array([101, 102, 103, 104, 105])
heart_rates = np.array([72, 110, 95, 130, 88])


status = np.where(heart_rates>100, "critical", "stable")

print("pationt_ID: ", pationt_ID)
print("heart_rates: ", heart_rates)
print("status: ", status)
```

```python
temp = np.array([36.5, 37.2, 0, 38.1, 0, 36.8])

condition = temp != 0

filtered = temp[condition]
print(filtered)
avg_temp = filtered.mean()
print(avg_temp)

cleaned_temp = np.where(temp == 0, avg_temp, temp)

print("original: ", temp)
print("clened: ", cleaned_temp)
```

```python
ages = np.array([25, 65, 70, 50, 45])
bp = np.array([120, 150, 135, 160, 145])

risk = np.where((( ages > 70) | (bp > 130) ), "risk", "normal" )

print(risk)
```

```python
salaries = np.array([400, 600, 550, 300, 700])

bonus = np.where((salaries>=500), salaries+100 , salaries+50)

print(bonus)
```

```python
marks = np.array([90, 75, 60, 82, 68])

grades = np.where(marks>= 85, "A", np.where(marks>=70, "B", "C"))

print(grades)
```

```python
marks = np.array([99, 75, 60, 82, 68, 89])

grades = np.where(marks>= 85, np.where(marks>= 95, "A+","A"), np.where(marks>=70, "B", "C"))

print(grades)
```

```python
purchases = np.array([50, 120, 80, 200, 150])

dis = np.where(purchases>100 , purchases*0.8, purchases*0.9)

print(dis)
```

```python
sales = pd.Series([200, 450, 300, 150, 500],  index=["Mon", "Tue", "Wed", "Thu", "Fri"])

# Total sales
print("Total:", sales.sum())

# Average sales
print("Average:", sales.mean())

# Max sales day
print("Highest day:", sales.idxmax())

# Sales > 300
print("Days with sales > 300:")
print(sales[sales > 300])
```
