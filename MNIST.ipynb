```python
# ==============================
# MNIST Multi-Class Classification
# Deep Learning Model
# ==============================

# Import Libraries
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten
from tensorflow.keras.optimizers import Adam
import matplotlib.pyplot as plt
import numpy as np


# ==============================
# 1. Load Dataset
# ==============================

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training Data:", x_train.shape)
print("Testing Data:", x_test.shape)


# ==============================
# 2. Data Normalization
# ==============================

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0


# ==============================
# 3. Build Deep Learning Model
# ==============================

model = Sequential([

    # Convert 28x28 image to vector
    Flatten(input_shape=(28,28)),

    # Hidden Layer 1
    Dense(256, activation='relu'),
    Dropout(0.3),

    # Hidden Layer 2
    Dense(128, activation='relu'),
    Dropout(0.2),

    # Hidden Layer 3
    Dense(64, activation='relu'),

    # Output Layer (10 classes)
    Dense(10, activation='softmax')
])


# ==============================
# 4. Compile Model
# ==============================

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)


# Model Architecture
model.summary()


# ==============================
# 5. Train Model
# ==============================

history = model.fit(
    x_train,
    y_train,
    epochs=20,
    batch_size=64,
    validation_split=0.2,
    verbose=1
)


# ==============================
# 6. Evaluate Model
# ==============================

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\nTest Accuracy:", test_accuracy)
print("Test Loss:", test_loss)


# ==============================
# 7. Plot Accuracy
# ==============================

plt.figure(figsize=(8,5))

plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Model Accuracy")

plt.legend()
plt.grid()
plt.show()


# ==============================
# 8. Plot Loss
# ==============================

plt.figure(figsize=(8,5))

plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Model Loss")

plt.legend()
plt.grid()
plt.show()


# ==============================
# 9. Prediction Example
# ==============================

index = 20

plt.imshow(x_test[index], cmap='gray')
plt.title("Actual Label: " + str(y_test[index]))
plt.axis("off")
plt.show()


prediction = model.predict(
    x_test[index].reshape(1,28,28)
)

predicted_class = np.argmax(prediction)

print("Predicted Class:", predicted_class)
print("Actual Class:", y_test[index])
```

```python
# ======================================
# MNIST Multi-Class Classification
# Deep Neural Network (ANN) - No CNN
# ======================================


# Import Libraries
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import matplotlib.pyplot as plt
import numpy as np


# ======================================
# 1. Load Dataset
# ======================================

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training Data:", x_train.shape)
print("Testing Data:", x_test.shape)


# ======================================
# 2. Normalization
# ======================================

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0



# ======================================
# 3. Build ANN Model
# ======================================

model = Sequential([

    # Convert 28x28 image to 784 features
    Flatten(input_shape=(28,28)),

    # Hidden Layer 1
    Dense(512, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),


    # Hidden Layer 2
    Dense(256, activation='relu'),
    BatchNormalization(),
    Dropout(0.25),


    # Hidden Layer 3
    Dense(128, activation='relu'),
    BatchNormalization(),
    Dropout(0.2),


    # Hidden Layer 4
    Dense(64, activation='relu'),


    # Output Layer
    Dense(10, activation='softmax')
])


# ======================================
# 4. Compile Model
# ======================================

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)


model.summary()



# ======================================
# 5. Callbacks
# ======================================

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=7,
    restore_best_weights=True
)


reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=3,
    min_lr=0.00001
)



# ======================================
# 6. Train Model
# ======================================

history = model.fit(

    x_train,
    y_train,

    epochs=50,
    batch_size=64,

    validation_split=0.2,

    callbacks=[
        early_stop,
        reduce_lr
    ],

    verbose=1
)



# ======================================
# 7. Evaluate Model
# ======================================

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)


print("\n======================")
print("Test Accuracy:", test_accuracy)
print("Test Loss:", test_loss)
print("======================")



# ======================================
# 8. Accuracy Plot
# ======================================

plt.figure(figsize=(8,5))

plt.plot(
    history.history['accuracy'],
    label='Training Accuracy'
)

plt.plot(
    history.history['val_accuracy'],
    label='Validation Accuracy'
)


plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Model Accuracy")

plt.legend()
plt.grid()
plt.show()



# ======================================
# 9. Loss Plot
# ======================================

plt.figure(figsize=(8,5))


plt.plot(
    history.history['loss'],
    label='Training Loss'
)

plt.plot(
    history.history['val_loss'],
    label='Validation Loss'
)


plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Model Loss")

plt.legend()
plt.grid()
plt.show()



# ======================================
# 10. Prediction Example
# ======================================

index = 20


plt.imshow(
    x_test[index],
    cmap='gray'
)

plt.title(
    "Actual Label: " + str(y_test[index])
)

plt.axis("off")
plt.show()



prediction = model.predict(
    x_test[index].reshape(1,28,28)
)


predicted_class = np.argmax(prediction)


print("Predicted Class:", predicted_class)
print("Actual Class:", y_test[index])
```
