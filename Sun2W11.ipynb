```python
import pandas as pd
import glob
import os
```

```python
path = "data/"
files = glob.glob(path + "/*.csv")

df_list = []

for file in files:
    df = pd.read_csv(file)
    df["Store"] = os.path.basename(file)
    df_list.append(df)

raw_df = pd.concat(df_list, ignore_index=True)
```

```python
print(raw_df.shape)
print(raw_df.columns)
raw_df.head()
```

```python
numeric_cols = raw_df.select_dtypes(include=["number"]).columns.tolist()

categorical_cols = raw_df.select_dtypes(
    include=["object", "string"]
).columns.tolist()
```

```python
X = raw_df[numeric_cols + categorical_cols]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])
```

```python
categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])
```

```python
preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_cols),
    ("cat", categorical_transformer, categorical_cols)
])
```

```python
X_transformed = preprocessor.fit_transform(X)
print(X_transformed.shape)
```

```python
encoded_cols = preprocessor.get_feature_names_out()

df_final = pd.DataFrame(X_transformed, columns=encoded_cols)
df_final.head()
```

```python
import pandas as pd

def reverse_one_hot(df, prefix):
    cols = [c for c in df.columns if c.startswith(prefix)]
    
    df[prefix] = df[cols].idxmax(axis=1).str.replace(prefix + "_", "")
    
    df = df.drop(columns=cols)
    return df
```
