```python
import re

dataset_raw = [
    "1. Sarah Johnson joined TechNova Solutions on 15 March 2025 as a software engineer. She relocated to Muscat, Oman, to work on artificial intelligence projects and collaborated with international teams to develop innovative solutions for the company.",
    "2. Dr. Ahmed Al Balushi attended the AI Summit organized by Future Vision Labs on January 12, 2026, in Dubai. During the event, he presented his research on natural language processing and participated in discussions with experts from around the world.",
    "3. Emily Carter signed a research partnership agreement with Global Health Institute on 8 August 2024 in London. The collaboration focused on developing machine learning models for disease prediction and improving healthcare decision-making through data analysis.",
    "4. Michael Brown presented the annual sustainability report for Green Energy Corporation on 30 September 2025 at the company's headquarters in New York City. The presentation highlighted renewable energy initiatives, environmental achievements, and future expansion plans across multiple regions.",
    "5. Fatima Al Hinai completed her internship at Digital Future Academy on 5 February 2026 in Salalah, Oman. During her internship, she gained practical experience in data science, worked on real-world analytics projects, and received recognition for her outstanding performance."
]

dataset = [re.sub(r'^\d+\.\s*', '', text) for text in dataset_raw]

for i, text in enumerate(dataset):
    print(f"Document {i+1}: {text}")
```

```python
import nltk

try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    print("Downloading NLTK punkt tokenizer...")
    nltk.download('punkt')
except ImportError:
    print("NLTK is not installed. Installing NLTK...")
    !pip install nltk
    import nltk
    nltk.download('punkt')

try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    print("Downloading NLTK punkt_tab tokenizer...")
    nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize, sent_tokenize

tokenized_sentences = []
for doc in dataset:
    sentences = sent_tokenize(doc)
    doc_tokens = [word_tokenize(sent) for sent in sentences]
    tokenized_sentences.append(doc_tokens)

print("أول جملة من أول مستند بعد الـ tokenization:")
print(tokenized_sentences[0][0])
```

```python
date_pattern = r'\b(?:January|February|March|April|May|June|July|August|September|October|November|December|Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)\s+\d{1,2},\s+\d{4}|\d{1,2}\s+(?:January|February|March|April|May|June|July|August|September|October|November|December|Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)\s+\d{4}\b'

extracted_entities = {
    'PERSON': [],
    'ORGANIZATION': [],
    'LOCATION': [],
    'DATE': [],
    'TIME': []
}

for doc_idx, doc_text in enumerate(dataset):
    print(f"\nProcessing Document {doc_idx + 1}:")

    dates_found = re.findall(date_pattern, doc_text)
    if dates_found:
        print(f"  Dates Found: {dates_found}")
        extracted_entities['DATE'].extend(dates_found)
    else:
        print("  No dates found in this document.")

print("\nجميع التواريخ المستخرجة:")
print(extracted_entities['DATE'])
```

```python
expected = [
    [
        ("Sarah Johnson", "PERSON"),
        ("TechNova Solutions", "ORGANIZATION"),
        ("15 March 2025", "DATE"),
        ("Muscat", "LOCATION"),
        ("Oman", "LOCATION")
    ],
    [
        ("Ahmed Al Balushi", "PERSON"),
        ("Future Vision Labs", "ORGANIZATION"),
        ("January 12, 2026", "DATE"),
        ("Dubai", "LOCATION")
    ],
    [
        ("Emily Carter", "PERSON"),
        ("Global Health Institute", "ORGANIZATION"),
        ("8 August 2024", "DATE"),
        ("London", "LOCATION")
    ],
    [
        ("Michael Brown", "PERSON"),
        ("Green Energy Corporation", "ORGANIZATION"),
        ("30 September 2025", "DATE"),
        ("New York City", "LOCATION")
    ],
    [
        ("Fatima Al Hinai", "PERSON"),
        ("Digital Future Academy", "ORGANIZATION"),
        ("5 February 2026", "DATE"),
        ("Salalah", "LOCATION"),
        ("Oman", "LOCATION")
    ]
]

predicted = []
for doc_idx, doc_text in enumerate(dataset):
    current_doc_entities = []

    dates_found = re.findall(date_pattern, doc_text)
    for date_ent in dates_found:
        current_doc_entities.append((date_ent, "DATE"))

    times_found = re.findall(time_pattern, doc_text, re.IGNORECASE)
    for time_ent in times_found:
        current_doc_entities.append((time_ent, "TIME"))

    organizations_found = re.findall(organization_pattern, doc_text)
    for org_ent in organizations_found:
        current_doc_entities.append((org_ent, "ORGANIZATION"))

    locations_found = re.findall(location_pattern, doc_text)
    for loc_ent in locations_found:
        if loc_ent == "Muscat, Oman":
            current_doc_entities.append(("Muscat", "LOCATION"))
            current_doc_entities.append(("Oman", "LOCATION"))
        elif loc_ent == "Salalah, Oman":
            current_doc_entities.append(("Salalah", "LOCATION"))
            current_doc_entities.append(("Oman", "LOCATION"))
        else:
            current_doc_entities.append((loc_ent, "LOCATION"))

    for refined_person in extracted_entities['PERSON']:
        if re.search(r'\b' + re.escape(refined_person) + r'\b', doc_text):
            current_doc_entities.append((refined_person, "PERSON"))

    current_doc_entities.sort(key=lambda x: (x[0], x[1]))
    predicted.append(current_doc_entities)

def evaluate_ner(expected, predicted):

    total_tp = 0
    total_fp = 0
    total_fn = 0

    for exp_par, pred_par in zip(expected, predicted):

        exp_set = set(exp_par)
        pred_set = set(pred_par)

        tp = len(exp_set & pred_set)
        fp = len(pred_set - exp_set)
        fn = len(exp_set - pred_set)

        total_tp += tp
        total_fp += fp
        total_fn += fn

    precision = total_tp / (total_tp + total_fp) if (total_tp + total_fp) else 0
    recall = total_tp / (total_tp + total_fn) if (total_tp + total_fn) else 0
    accuracy = total_tp / sum(len(p) for p in expected)
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0

    print("\n===== NER MODEL EVALUATION =====\n")
    print(f"True Positives (TP): {total_tp}")
    print(f"False Positives (FP): {total_fp}")
    print(f"False Negatives (FN): {total_fn}")
    print("--------------------------------")
    print(f"Accuracy : {accuracy:.2%}")
    print(f"Precision: {precision:.2%}")
    print(f"Recall   : {recall:.2%}")
    print(f"F1 Score : {f1:.2%}")
    print("--------------------------------\n")

evaluate_ner(expected, predicted)
```

```python
new_dataset_raw = [
    "6. Dr. Ali Hassan attended a conference on AI ethics at Cairo University on 25 April 2024. He presented his research at 10:30 AM.",
    "7. Maria Lopez joined BioGen Corp in London on 10 January 2025 as a data scientist.",
    "8. The 'Future Tech Summit' will be held in Riyadh, Saudi Arabia, on 15 March 2026."
]

new_dataset = [re.sub(r'^\d+\.\s*', '', text) for text in new_dataset_raw]

print("\n--- New Dataset ---")
for i, text in enumerate(new_dataset):
    print(f"Document {i+1}: {text}")

new_extracted_entities = {
    'PERSON': [],
    'ORGANIZATION': [],
    'LOCATION': [],
    'DATE': [],
    'TIME': []
}

# Extract Dates
print("\n--- Extracting Dates from New Dataset ---")
for doc_idx, doc_text in enumerate(new_dataset):
    dates_found = re.findall(date_pattern, doc_text)
    if dates_found:
        new_extracted_entities['DATE'].extend(dates_found)
        print(f"  Document {doc_idx + 1} Dates: {dates_found}")

# Extract Times
print("\n--- Extracting Times from New Dataset ---")
for doc_idx, doc_text in enumerate(new_dataset):
    times_found = re.findall(time_pattern, doc_text, re.IGNORECASE)
    if times_found:
        new_extracted_entities['TIME'].extend(times_found)
        print(f"  Document {doc_idx + 1} Times: {times_found}")

# Extract Organizations
print("\n--- Extracting Organizations from New Dataset ---")
for doc_idx, doc_text in enumerate(new_dataset):
    organizations_found = re.findall(organization_pattern, doc_text)
    if organizations_found:
        new_extracted_entities['ORGANIZATION'].extend(organizations_found)
        print(f"  Document {doc_idx + 1} Organizations: {organizations_found}")

# Extract Locations
print("\n--- Extracting Locations from New Dataset ---")
for doc_idx, doc_text in enumerate(new_dataset):
    locations_found = re.findall(location_pattern, doc_text)
    if locations_found:
        new_extracted_entities['LOCATION'].extend(locations_found)
        print(f"  Document {doc_idx + 1} Locations: {locations_found}")

# Extract Persons (using the pattern and then filtering)
print("\n--- Extracting Persons from New Dataset ---")
for doc_idx, doc_text in enumerate(new_dataset):
    persons_found = re.findall(person_pattern, doc_text)
    if persons_found:
        # Filter out common non-person words that might match the pattern
        filtered_persons = [p.strip() for p in persons_found if p not in ['March', 'January', 'April', 'Future', 'Tech']]
        new_extracted_entities['PERSON'].extend(filtered_persons)
        print(f"  Document {doc_idx + 1} Persons (pre-filtered): {persons_found}")
        print(f"  Document {doc_idx + 1} Persons (filtered): {filtered_persons}")

# Refine persons similar to the previous step (deduplicate and filter based on other entity types)
new_refined_persons = []
for person_candidate in new_extracted_entities['PERSON']:
    is_organization = False
    is_location = False

    # Check if person_candidate is an organization or part of one
    # Check if person_candidate itself is in the known organization list
    if person_candidate in organization_names:
        is_organization = True
    if not is_organization: # Only if not already identified as a full organization
        for org in new_extracted_entities['ORGANIZATION']:
            if person_candidate in org: # Check if person candidate is a part of extracted organization
                is_organization = True
                break

    # Check if person_candidate is a location or part of one
    # First, check if the person_candidate itself is one of the known locations
    if person_candidate in location_names_sorted:
        is_location = True
    # Second, check if the person_candidate is a substring of an extracted location
    if not is_location: # Only if not already identified as a full location
        for loc in new_extracted_entities['LOCATION']:
            if person_candidate in loc: # Check if person candidate is a part of extracted location
                is_location = True
                break

    if not is_organization and not is_location:
        new_refined_persons.append(person_candidate.strip())

new_extracted_entities['PERSON'] = list(set(new_refined_persons))

# Deduplicate all other entity types
new_extracted_entities['ORGANIZATION'] = list(set(new_extracted_entities['ORGANIZATION']))
new_extracted_entities['LOCATION'] = list(set(new_extracted_entities['LOCATION']))
new_extracted_entities['DATE'] = list(set(new_extracted_entities['DATE']))
new_extracted_entities['TIME'] = list(set(new_extracted_entities['TIME']))

print("\n--- All Extracted Entities from New Dataset ---")
for entity_type, entities in new_extracted_entities.items():
    print(f"{entity_type}: {entities}")
```

```python
expected_new_dataset = [
    [
        ("Ali Hassan", "PERSON"),
        ("Cairo University", "ORGANIZATION"),
        ("25 April 2024", "DATE"),
        ("10:30 AM", "TIME")
    ],
    [
        ("Maria Lopez", "PERSON"),
        ("BioGen Corp", "ORGANIZATION"),
        ("London", "LOCATION"),
        ("10 January 2025", "DATE")
    ],
    [
        ("Future Tech Summit", "ORGANIZATION"),
        ("Riyadh, Saudi Arabia", "LOCATION"),
        ("15 March 2026", "DATE")
    ]
]

predicted_new_dataset = []
for doc_idx, doc_text in enumerate(new_dataset):
    current_doc_entities = []

    # Extract Dates
    dates_found = re.findall(date_pattern, doc_text)
    for date_ent in dates_found:
        current_doc_entities.append((date_ent, "DATE"))

    # Extract Times
    times_found = re.findall(time_pattern, doc_text, re.IGNORECASE)
    for time_ent in times_found:
        current_doc_entities.append((time_ent, "TIME"))

    # Extract Organizations first, as locations might be part of organizations
    organizations_found_in_doc = re.findall(organization_pattern, doc_text)
    for org_ent in organizations_found_in_doc:
        current_doc_entities.append((org_ent, "ORGANIZATION"))

    # Extract Locations, but filter out those that are part of an extracted organization
    locations_found = re.findall(location_pattern, doc_text)
    for loc_ent in locations_found:
        is_part_of_org = False
        for org_ent in organizations_found_in_doc:
            if loc_ent in org_ent:
                is_part_of_org = True
                break

        if not is_part_of_org:
            if loc_ent == "Muscat, Oman":
                current_doc_entities.append(("Muscat", "LOCATION"))
                current_doc_entities.append(("Oman", "LOCATION"))
            elif loc_ent == "Salalah, Oman":
                current_doc_entities.append(("Salalah", "LOCATION"))
                current_doc_entities.append(("Oman", "LOCATION"))
            else:
                current_doc_entities.append((loc_ent, "LOCATION"))

    # Extract Persons (using new_refined_persons from ONnB4WWZP33I)
    # We need to iterate over the new_refined_persons and check if they exist in the current document
    # This is a bit tricky as new_extracted_entities['PERSON'] is a flat list of unique persons,
    # not linked to specific documents. We'll simulate by checking presence in doc_text
    for refined_person in new_extracted_entities['PERSON']:
        # Ensure the person is actually in the current document to avoid false positives across documents
        if re.search(r'\b' + re.escape(refined_person) + r'\b', doc_text):
            current_doc_entities.append((refined_person, "PERSON"))

    current_doc_entities.sort(key=lambda x: (x[0], x[1]))
    predicted_new_dataset.append(current_doc_entities)

# Define the evaluate_ner function again for self-containment in this cell
def evaluate_ner(expected, predicted):
    total_tp = 0
    total_fp = 0
    total_fn = 0

    # Lists to store detailed FPs and FNs per document
    detailed_fps = [set() for _ in range(len(expected))]
    detailed_fns = [set() for _ in range(len(expected))]

    for doc_idx, (exp_par, pred_par) in enumerate(zip(expected, predicted)):
        exp_set = set(exp_par)
        pred_set = set(pred_par)

        tp = len(exp_set & pred_set)
        fp = len(pred_set - exp_set)
        fn = len(exp_set - pred_set)

        total_tp += tp
        total_fp += fp
        total_fn += fn

        detailed_fps[doc_idx] = pred_set - exp_set
        detailed_fns[doc_idx] = exp_set - pred_set

    precision = total_tp / (total_tp + total_fp) if (total_tp + total_fp) else 0
    recall = total_tp / (total_tp + total_fn) if (total_tp + total_fn) else 0
    accuracy = total_tp / sum(len(p) for p in expected) if sum(len(p) for p in expected) else 0 # Handle division by zero
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0

    print("\n===== NER MODEL EVALUATION (NEW DATASET) =====\n")
    print(f"True Positives (TP): {total_tp}")
    print(f"False Positives (FP): {total_fp}")
    print(f"False Negatives (FN): {total_fn}")
    print("--------------------------------")
    print(f"Accuracy : {accuracy:.2%}")
    print(f"Precision: {precision:.2%}")
    print(f"Recall   : {recall:.2%}")
    print(f"F1 Score : {f1:.2%}")
    print("--------------------------------\n")

    print("\n--- تحليل الأخطاء التفصيلي للبيانات الجديدة (Detailed Error Analysis for New Dataset) ---")
    for doc_idx in range(len(expected)):
        print(f"\nDocument {doc_idx + 1}:")
        if detailed_fps[doc_idx]:
            print(f"  False Positives (تم استخلاصها خطأ): {list(detailed_fps[doc_idx])}")
        else:
            print("  No False Positives in this document.")

        if detailed_fns[doc_idx]:
            print(f"  False Negatives (لم يتم استخلاصها): {list(detailed_fns[doc_idx])}")
        else:
            print("  No False Negatives in this document.")

evaluate_ner(expected_new_dataset, predicted_new_dataset)
```
