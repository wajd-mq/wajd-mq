```python
import pandas as pd

url = "https://huggingface.co/datasets/stanfordnlp/imdb/resolve/main/plain_text/train-00000-of-00001.parquet"

df = pd.read_parquet(url)

print(df.head())
```

```python
df["review_length"] = df["text"].apply(len)
```

```python
import matplotlib.pyplot as plt

plt.hist(df["review_length"], bins=30)
plt.title("Original Distribution")
plt.show()
```

```python
import numpy as np

sample_means = []

for i in range(1000):
    sample = df["review_length"].sample(50, replace=True)
    sample_means.append(sample.mean())
```

```python
plt.hist(sample_means, bins=30)
plt.title("Sampling Distribution (CLT)")
plt.show()
```

```python
print("Original Mean:", df["review_length"].mean())
print("Sample Means Mean:", np.mean(sample_means))
```
